# Model evaluation

Thin wrapper around `cfb_model.evaluate`. Train Ridge + XGBoost on leakage-safe as-of features and compare to Elo and the closing spread.

```bash
python -m cfb_model migrate   # or ingest
python -m cfb_model evaluate
```

````python
from pathlib import Path
import json
import pandas as pd

from cfb_model.evaluate import walk_forward, write_metrics
from cfb_model.features import build_features
from cfb_model.models import ridge_coefficients, save_models, xgb_importances
````

````python
frame = build_features()
print(frame.shape)
frame.head()
````

````python
scored, summary, models = walk_forward(frame)
save_models(models)
write_metrics(summary)
pd.DataFrame(summary["folds"])
````

````python
print(json.dumps(summary["overall"], indent=2))
ridge_coefficients(models).head(15)
````

````python
xgb_importances(models).head(15)
````

````python
ax = scored.plot.scatter(x="pred_margin", y="home_margin", alpha=0.35, figsize=(6, 6))
ax.axline((0, 0), slope=1, color="black", linewidth=1)
ax.set_title("Predicted vs actual home margin")
````
